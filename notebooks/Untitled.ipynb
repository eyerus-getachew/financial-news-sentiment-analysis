# Exploratory Data Analysis

This notebook explores the financial news dataset to identify:
- textual patterns
- publisher activity
- publication trends
- recurring financial themes

```python
df['headline_length'] = df['headline'].apply(len)

df['headline_length'].describe()
```

```python
sns.histplot(df['headline_length'], bins=50)

plt.title("Headline Length Distribution")
plt.xlabel("Headline Length")
plt.ylabel("Frequency")

plt.show()
```

Most headlines are relatively short and concise, which is typical for financial reporting headlines designed to communicate key market information quickly.
